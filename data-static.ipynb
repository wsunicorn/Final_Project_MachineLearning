```python
import pandas as pd
# === Hàm check nhãn từ tên file
def label_check(filename):
    filename = filename.lower()
    if 'fake' in filename:
        return 'fake'
    elif 'real' in filename:
        return 'real'
    else:
        return -1
```

```python
df = pd.read_csv('/kaggle/input/data-img-fake-real/labels.csv',header=None)
df = df.rename(columns = {0:'filename'})
df.head()
```

```python
df['labels'] = df['filename'].apply(label_check)
df['labels'].value_counts()
```

# Static - real

```python
# Hàm trích loại cảnh vật cụ thể từ tên file: jungle, mountain, sea, beach,...
df_real = df[df['labels'] == 'real']

def extract_scene_type(filename):
    filename = filename.lower()
    scene_keywords = ['jungle', 'mountain', 'sea', 'beach', 'forest', 'desert', 'city', 'river', 'lake', 'snow']

    for word in scene_keywords:
        if word in filename:
            return word
    return 'unknown'

# Áp dụng hàm trích loại cảnh vật vào nhóm ảnh real
df_real['scene_type'] = df_real[df_real.columns[0]].apply(extract_scene_type)

# Đếm số lượng từng loại cảnh vật trong nhóm real
df_real['scene_type'].value_counts().reset_index()

```

```python
# Lọc ra các ảnh real có scene_type là unknown
unknown_real_images = df_real[df_real['scene_type'] == 'unknown']

# Hiển thị danh sách tên ảnh
unknown_real_images[df_real.columns[0]].reset_index(drop=True)


```

```python
# Cập nhật hàm với khả năng sửa lỗi chính tả nhẹ

def extract_scene_type_v2(filename):
    filename = filename.lower()
    # Danh sách từ khóa đúng
    scene_keywords = ['jungle', 'mountain', 'sea', 'beach', 'forest', 'desert', 'city', 'river', 'lake', 'snow']
    typo_corrections = {
        'see': 'sea',
        'junlge': 'jungle'
    }

    # Sửa lỗi chính tả trong tên file
    for typo, correct in typo_corrections.items():
        filename = filename.replace(typo, correct)

    # Tìm từ cảnh vật
    for word in scene_keywords:
        if word in filename:
            return word
    return 'unknown'

df_real['scene_type'] = df_real[df_real.columns[0]].apply(extract_scene_type_v2)

df_real['scene_type'].value_counts().reset_index()

```

# Static - Fake

```python
df_fake = df[df['labels'] == 'fake']

df_fake['scene_type'] = df_fake[df_fake.columns[0]].apply(extract_scene_type_v2)

df_fake['scene_type'].value_counts().reset_index()

```

```python
# Hàm trích công cụ tạo ảnh từ tên file (ví dụ: stable, dalle, dreamstudio...)
def extract_tool(filename):
    filename = filename.lower()
    tool_keywords = ['stable', 'dalle', 'dall.e', 'dreamstudio', 'midjourney', 'bing', 'firefly']

    for tool in tool_keywords:
        if tool in filename:
            return tool
    return 'unknown'

df_fake['tool'] = df_fake[df_fake.columns[0]].apply(extract_tool)

df_fake['tool'].value_counts().reset_index()

```

```python
unknown_fake_images = df_fake[df_fake['tool'] == 'unknown']

unknown_fake_images[df_fake.columns[0]].reset_index(drop=True)

```

```python
# Hàm cập nhật với các biến thể mới để xác định công cụ tạo ảnh
def extract_tool_v2(filename):
    filename = filename.lower()
    tool_keywords = {
        'stable': 'stable',
        'dalle': 'dalle',
        'dall.e': 'dalle',
        'delle': 'dalle',
        'dall': 'dalle',
        'dreamstudio': 'dreamstudio',
        'dream': 'dreamstudio',
        'dreamai': 'dreamstudio',
        'craiyon': 'craiyon',
        'midjourney': 'midjourney',
        'bing': 'bing',
        'firefly': 'firefly'
    }

    for key in tool_keywords:
        if key in filename:
            return tool_keywords[key]
    return 'unknown'

df_fake['tool'] = df_fake[df_fake.columns[0]].apply(extract_tool_v2)

df_fake['tool'].value_counts().reset_index()

```
